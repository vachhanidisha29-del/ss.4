```python
import numpy as np

steps = np.array([4000, 5500, 6000, 4500, 7000, 8000, 6500])

# Add 500 bonus steps to each day using broadcasting
updated_steps = steps + 500

# Calculate total steps for the week
total_steps = np.sum(updated_steps)

updated_steps, total_steps
```
